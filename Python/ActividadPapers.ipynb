## 2. Resumen del artículo

El capítulo titulado *"Empowering Scientific Computing and Data Manipulation With Numerical Python (NumPy)"* examina a fondo el rol de NumPy en el ecosistema de Python para la computación científica y el análisis de datos. El artículo, escrito por Tesfaye Fufa Gedefa, Galety Mohammed Gouse y Garamu Tilahun Iticha (2023), destaca cómo NumPy se ha convertido en la piedra angular para el desarrollo de aplicaciones numéricas y científicas, proporcionando estructuras de datos eficientes y herramientas robustas para el tratamiento de datos de múltiples dimensiones.

### Principales ideas desarrolladas:

#### I. El objeto `ndarray`
El núcleo de NumPy es el objeto `ndarray`, una estructura de datos N-dimensional que permite representar y manipular arrays de forma eficiente. Este objeto admite operaciones vectorizadas, es decir, operaciones que se aplican a colecciones completas de datos sin necesidad de bucles explícitos.

#### II. Tipos de datos en NumPy
NumPy admite distintos tipos de datos numéricos: enteros sin signo (`uint8`, `uint16`), enteros con signo (`int8`, `int16`), números en coma flotante (`float32`, `float64`), números complejos, y también tipos personalizados. A través del `dtype`, los arrays pueden configurarse para ser altamente eficientes en memoria y procesamiento.

#### III. Operaciones vectorizadas y rendimiento
Las operaciones aritméticas básicas (suma, resta, multiplicación, división), así como operaciones más complejas (algebra lineal, transformadas, estadísticas), están optimizadas para ejecutarse rápidamente gracias a la asignación contigua de memoria y el uso de instrucciones SIMD (Single Instruction Multiple Data).

#### IV. Diferencias entre listas y arrays
A diferencia de las listas de Python, los arrays de NumPy:
- Ocupan menos memoria.
- Permiten procesamiento numérico más rápido.
- Son homogéneos (todos sus elementos tienen el mismo tipo de dato).
- Permiten operaciones matemáticas nativas sin necesidad de iteración explícita.

#### V. Técnicas clave: slicing, broadcasting, reshaping
El artículo explica conceptos avanzados como:
- **Slicing**: selección eficiente de subconjuntos de arrays.
- **Broadcasting**: operaciones entre arrays de diferentes formas, mediante reglas internas que "extienden" sus dimensiones.
- **Reshaping**: transformación de la forma (shape) del array sin alterar sus datos.

#### VI. Funciones estadísticas y algebraicas
NumPy proporciona funciones para:
- Análisis estadístico (mínimo, máximo, media, desviación estándar...).
- Álgebra lineal (producto de matrices, determinantes, sistemas de ecuaciones...).
- Trigonometría, redondeo, búsqueda y ordenación.

#### VII. Aplicaciones e interoperabilidad
NumPy es utilizado ampliamente en ciencia de datos, ingeniería, redes complejas, machine learning y big data. Además, se integra fácilmente con otros lenguajes como C/C++ y con bibliotecas como Pandas, SciPy o TensorFlow.

## 4. Tipos de datos en NumpPy

NumPy trabaja principalmente con un tipo de estructura de datos llamada `ndarray` (N-dimensional array), que puede representar desde un número escalar hasta un tensor de múltiples dimensiones. Los tipos más comunes que podemos encontrar son:

- **Escalar (0D)**  
  Un único valor numérico, sin dimensiones.

- **Vector (1D)**  
  Una lista unidimensional de valores.

- **Matriz (2D)**  
  Una tabla de valores con filas y columnas (array bidimensional).

- **Tensor (nD, n > 2)**  
  Una estructura de datos multidimensional (3D o más), muy útil en aplicaciones científicas y de inteligencia artificial.

A continuación se muestran ejemplos en Python para cada uno de estos tipos utilizando la función `np.array()`.


```python
# Escalar: una sola cantidad (0 dimensiones)
escalar = np.array(7)
print("Escalar:", escalar)
print("Dimensiones:", escalar.ndim)
print("Forma:", escalar.shape)
```

```python
# Vector: una lista de valores (1D)
vector = np.array([10, 20, 30])
print("Vector:", vector)
print("Dimensiones:", vector.ndim)
print("Forma:", vector.shape)
```

```python
# Matriz: array bidimensional (2D)
matriz = np.array([[1, 2, 3],
                   [4, 5, 6]])
print("Matriz:\n", matriz)
print("Dimensiones:", matriz.ndim)
print("Forma:", matriz.shape)
```

```python
# Tensor: array tridimensional (3D)
tensor = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
print("Tensor:\n", tensor)
print("Dimensiones:", tensor.ndim)
print("Forma:", tensor.shape)
```

## 5. Operaciones con Arrays

NumPy permite realizar operaciones matemáticas y estructurales sobre arrays de forma sencilla, rápida y vectorizada, es decir, sin necesidad de usar bucles.

A continuación se presentan algunas de las operaciones más comunes:

- **Operaciones aritméticas**: suma, resta, multiplicación, división, etc.
- **Transposición de arrays**: invertir filas por columnas.
- **Concatenación y división**: unir o dividir arrays.
- **Redimensionamiento (reshape)**: cambiar la forma del array sin alterar los datos.


```python
# Creamos dos arrays
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Operaciones elemento a elemento
print("Suma:", a + b)
print("Resta:", a - b)
print("Multiplicación:", a * b)
print("División:", a / b)
```

```python
# Matriz 2D
matriz = np.array([[1, 2, 3],
                   [4, 5, 6]])

# Transponer (intercambiar filas y columnas)
transpuesta = matriz.T

print("Matriz original:\n", matriz)
print("Transpuesta:\n", transpuesta)
```

```python
# Concatenar dos vectores horizontalmente
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])
concat = np.concatenate([x, y])

print("Concatenación:", concat)
```

```python
# Dividir un array en tres partes iguales
z = np.array([10, 20, 30, 40, 50, 60])
split = np.split(z, 3)

print("División en partes:")
for i, part in enumerate(split):
    print(f"Parte {i+1}:", part)
```

```python
# Reshape de un array de 1D a 2D
original = np.array([1, 2, 3, 4, 5, 6])
reshape = original.reshape((2, 3))

print("Array original:", original)
print("Reshape a 2x3:\n", reshape)
```

## 6. Slicing, Indexing y Broadcasting

NumPy proporciona herramientas avanzadas para acceder, modificar y operar sobre subconjuntos de arrays de forma eficiente. Estas son claves para trabajar con datos a gran escala.

- **Indexing**: acceder a elementos concretos de un array mediante índices.
- **Slicing**: seleccionar rangos de elementos utilizando la notación `[inicio:fin:paso]`.
- **Broadcasting**: permite realizar operaciones aritméticas entre arrays de distintas formas (dimensiones), siempre que sean compatibles bajo las reglas de broadcasting.

Estas capacidades permiten escribir código más conciso, rápido y expresivo.

```python
# Crear un array 2D
matriz = np.array([[10, 20, 30],
                   [40, 50, 60],
                   [70, 80, 90]])

# Acceder al elemento de la fila 1, columna 2
print("Elemento [1,2]:", matriz[1, 2])  # Output: 60
```

```python
# Seleccionar un subconjunto de filas y columnas
submatriz = matriz[0:2, 1:]  # Filas 0-1, columnas 1-2
print("Submatriz con slicing:\n", submatriz)

# Slicing en 1D
vector = np.array([5, 10, 15, 20, 25, 30])
print("Segmento vector[1:5:2]:", vector[1:5:2])  # Output: [10 20]
```

```python
# Array 2D
a = np.array([[1, 2, 3],
              [4, 5, 6]])

# Vector 1D que se "extiende" automáticamente (broadcast)
b = np.array([10, 20, 30])

# Suma por filas (cada elemento de b se suma a una fila de a)
resultado = a + b

print("Resultado de broadcasting:\n", resultado)
```

```python
# Multiplicación por escalar (broadcast implícito)
x = np.array([1, 2, 3, 4])
print("Multiplicación por escalar (2):", x * 2)
```

## 7. Funciones especiales de Numpy

Además de operaciones básicas, NumPy proporciona una amplia gama de funciones especializadas para el cálculo matemático, estadístico y científico. A continuación se presentan algunas de las más relevantes:

- **Funciones trigonométricas**: cálculos con ángulos en radianes.
- **Funciones de redondeo**: aproximación de valores decimales.
- **Funciones estadísticas**: análisis de datos (mínimo, máximo, media, etc.).
- **Álgebra lineal**: operaciones con matrices, determinantes, autovalores, etc.

Estas funciones hacen de NumPy una herramienta clave en análisis numérico y modelado matemático.

```python
# Ángulos en radianes
ángulos = np.array([0, np.pi/2, np.pi])

# Funciones trigonométricas
print("Seno:", np.sin(ángulos))
print("Coseno:", np.cos(ángulos))
print("Tangente:", np.tan(ángulos))
```

```python
# Array con decimales
valores = np.array([3.14159, 2.71828, 1.61803])

# Redondear a 2 decimales
print("Original:", valores)
print("Redondeado (2 decimales):", np.round(valores, 2))
```

```python
# Datos simulados
datos = np.array([10, 20, 30, 40, 50])

print("Mínimo:", np.min(datos))
print("Máximo:", np.max(datos))
print("Media:", np.mean(datos))
print("Desviación estándar:", np.std(datos))
```

```python
# Matriz 2x2
A = np.array([[3, 1],
              [2, 4]])

# Vector columna
b = np.array([9, 10])

# Resolver sistema lineal Ax = b
x = np.linalg.solve(A, b)
print("Solución del sistema Ax = b:", x)

# Determinante
print("Determinante de A:", np.linalg.det(A))

# Matriz inversa
print("Inversa de A:\n", np.linalg.inv(A))
```

## 8. Diferencias listas y arrays

Aunque las listas de Python y los arrays de NumPy pueden parecer similares, existen diferencias clave que hacen que los arrays sean más eficientes y adecuados para el procesamiento numérico:

| Característica              | Lista de Python             | Array de NumPy                   |
|----------------------------|-----------------------------|----------------------------------|
| Tipos de datos             | Heterogéneos                | Homogéneos                       |
| Velocidad                  | Más lenta                   | Más rápida (vectorización)       |
| Consumo de memoria         | Mayor                       | Menor                            |
| Operaciones matemáticas    | Requiere bucles manuales    | Nativas y optimizadas            |
| Necesidad de importar      | No                          | Sí (`import numpy as np`)        |

**Conclusión:**  
Las listas son estructuras versátiles para datos heterogéneos y generales, mientras que los arrays de NumPy están optimizados para cálculos numéricos a gran escala, siendo mucho más rápidos y eficientes.

```python
import numpy as np
import sys
import time

# Tamaño de muestra
n = 1_000_000

# Lista de Python
lista = list(range(n))

# Array de NumPy
array = np.arange(n)

# Memoria usada
print("Memoria ocupada por la lista:", sys.getsizeof(lista), "bytes")
print("Memoria ocupada por el array:", array.nbytes, "bytes")

# Suma: lista vs array
start = time.time()
lista_resultado = [x + 1 for x in lista]
print("Tiempo de suma con lista:", round(time.time() - start, 4), "s")

start = time.time()
array_resultado = array + 1
print("Tiempo de suma con array:", round(time.time() - start, 4), "s")
```

## 9. Aplicación práctica

Una aplicación habitual en ciencia de datos es la **normalización**, que transforma un conjunto de datos para que sus valores estén en un rango específico (habitualmente entre 0 y 1). Esto permite mejorar la estabilidad numérica de los algoritmos y evita que una escala domine a otra.

### - Ejemplo: Normalización Min-Max

Dado un array de números enteros aleatorios, aplicaremos la fórmula:

    [X_{norm} = \frac{X - X_{min}}{X_{max} - X_{min}}]

Este proceso se realiza fácilmente gracias a la capacidad de operaciones vectorizadas de NumPy.

```python
# Generar datos aleatorios
datos = np.random.randint(1, 1000, size=10)
print("Datos originales:", datos)

# Normalización min-max
normalizado = (datos - np.min(datos)) / (np.max(datos) - np.min(datos))
print("Datos normalizados:", normalizado)
```
